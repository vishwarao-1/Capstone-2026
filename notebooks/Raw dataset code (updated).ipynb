```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("cumulative.csv",encoding='utf-8',nrows=9564)
print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.head()
df.info()
df.describe()
df.shape

print(df.head())
print(df.info())
print(df.describe())
print(df.shape)

print("Column names:\n")
for column in df.columns:
    print(column)


print("--------TO FIND EMPTY COLUMNS--------")
emp_col = []
for column in df.columns:
    if df[column].isnull().all():
        emp_col.append(column)
print("Completely empty columns:")
print(emp_col)

print("Number of completely empty columns:", len(emp_col))



print("-----MISSING VALUES-----")
missing_values = df.isnull().sum()
print(missing_values)


print("------MISSING VALUES PERCENTAGE-------")
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage.round(2))



print("------MISSING REPORT------")
missing_report = pd.DataFrame()
missing_report["Column"] = df.columns
missing_report["Missing Values"] = df.isnull().sum().values
missing_report["Missing Percentage"] = (
    df.isnull().sum().values / len(df) * 100
).round(2)
print(missing_report)



missing_report = missing_report.sort_values(
    "Missing Percentage",
    ascending=False
)
print(missing_report)



print("-----DUPLICATE ROWS------")
dupl_rows = df.duplicated().sum()
print("Number of duplicate rows:", dupl_rows)


if dupl_rows > 0:
    df[df.duplicated()]
else:
    print("No duplicate rows found.")


dupl_kepoi = df["kepoi_name"].duplicated().sum()
print("Duplicate KOI names:", dupl_kepoi)

df[df["kepoi_name"].duplicated(keep=False)][
    ["kepoi_name", "koi_disposition"]
].head(20)

print(df["koi_disposition"].value_counts(dropna=False))





print("-----Numerical columns-----")
num_columns = df.select_dtypes(
    include=np.number
).columns
print("Number of numerical columns:", len(num_columns))
print("\nNumerical columns:")
for column in num_columns:
    print(column)



print("-------Catergorical columns-------")
cate_columns = df.select_dtypes(
    include="object"
).columns
print("Number of categorical columns:", len(cate_columns))
print("\nCategorical columns:")
for column in cate_columns:
    print(column)


print("-----check infinity values-------------")
infi_values = np.isinf(df[num_columns]).sum()
print("Columns with infinite values:")
print(infi_values[infi_values > 0])
print()


print("-----Check negative values-----------")
neg_values = (df[num_columns] < 0).sum()
print("Columns with negative values:")
print(neg_values[neg_values > 0])
print()
















print("--Datasets--")
df_clean = df.copy()
print("Original dataset:", df.shape)
print("Cleaning dataset:", df_clean.shape)
print()


print("--New dataset--")
df_clean = df_clean.drop(columns=emp_col)
print("Removed empty columns:", len(emp_col))
print("New dataset shape:", df_clean.shape)
print()



print("--column details--")
print("Original columns:", len(df.columns))
print("Cleaned columns:", len(df_clean.columns))
print()



print("--Remaining complete empty col--")
print("Remaining completely empty columns:")
print(
    df_clean.columns[
        df_clean.isnull().all()
    ].tolist()
)

for column in df_clean.columns:
    print(column, "→", df_clean[column].dtype)
print()




print("--ID for unique column--")
id_col = [
    "kepid",
    "kepoi_name",
    "kepler_name"
]
target_col = "koi_disposition"

print("ID columns:")
print(id_col)
print()
print("\nTarget column:")
print(target_col)
print()




print("-----------------------")
for column in id_col:
    if column in df_clean.columns:
        print(column, " found")
    else:
        print(column, " Not found")
print()



print("------------------------")
cate_columns = df_clean.select_dtypes(
    include="object"
).columns.tolist()

print("Categorical columns:")
for column in cate_columns:
    print("-", column)
print()



print("------------------------")
num_columns = df_clean.select_dtypes(
    include=np.number
).columns.tolist()

print("Numerical columns:")
for column in num_columns:
    print("-", column)
print()




print("-------------------------")
print("Target column:", target_col)
print("\nValues:")
print(df_clean[target_col].value_counts(dropna=False))
print()





print("------------------------------")
col_type = []

for col in df_clean.columns:
    if col in id_col:
        category = "ID"
    elif col == target_col:
        category = "Target"
    elif col in cate_columns:
        category = "Categorical"
    else:
        category = "Numerical"
    col_type.append(category)

col_clasfic = pd.DataFrame({
    "Column": df_clean.columns,
    "Data Type": df_clean.dtypes.astype(str).values,
    "Category": col_type
})
print(col_clasfic)
print()




print("-----------------------")
col_clasfic["Missing Values"] = (
    df_clean.isnull().sum().values
)
col_clasfic["Missing %"] = (
    df_clean.isnull().sum().values /
    len(df_clean) * 100
).round(2)
print(col_clasfic)
print()





print("----------High missing columns--------------")
high_mis = col_clasfic[
    col_clasfic["Missing %"] >= 50
]
print("Columns with 50% or more missing values:")
display(high_mis)
print()






print("-------------Low missing columns-----------")
low_mis = col_clasfic[
    (col_clasfic["Missing %"] > 0) &
    (col_clasfic["Missing %"] < 50)
]
print("Columns with less than 50% missing values:")
display(low_mis)
print()






print("----------------------------")
col_clasfic.to_csv(
    "column_classification.csv",
    index=False
)
print("Column classification saved successfully.")
print()
print()





# Creating data dictionary for the reference of our project
#Data dictionary separate csv file will provide

print("------Data dict-----------")
data_dic= pd.DataFrame()
data_dic["Column"] = df_clean.columns
data_dic["Data Type"] = df_clean.dtypes.astype(str).values
print(data_dic)
print()



print("------After catergory added in data dic-----------")
data_dic["Category"] = col_clasfic["Category"].values
print(data_dic)
print()



print("--------------------------------------------------------")
data_dic["Missing Values"] = df_clean.isnull().sum().values
data_dic["Missing %"] = (
    df_clean.isnull().sum().values / len(df_clean) * 100
).round(2)
print(data_dic)
print()




# Meanings or full form of the keplers column name in the given project
print("------------------------------------------------------------------")
kepler_meanings = {
    "kepid": "Kepler identification number",
    "kepoi_name": "Kepler Object of Interest name",
    "kepler_name": "Official Kepler name assigned to a confirmed planet",
    "koi_disposition": "Final classification/disposition of the object",
    "koi_period": "Orbital period of the candidate",
    "koi_period_err1": "Upper uncertainty of orbital period",
    "koi_period_err2": "Lower uncertainty of orbital period",
    "koi_time0bk": "Transit midpoint time",
    "koi_time0bk_err1": "Upper uncertainty of transit midpoint",
    "koi_time0bk_err2": "Lower uncertainty of transit midpoint",
    "koi_impact": "Impact parameter of the transit",
    "koi_depth": "Transit depth",
    "koi_prad": "Estimated planetary radius",
    "koi_teq": "Estimated equilibrium temperature",
    "koi_insol": "Estimated incident stellar flux",
    "koi_steff": "Effective temperature of the host star",
    "koi_slogg": "Surface gravity of the host star",
    "koi_srad": "Radius of the host star"
}

data_dic["Meaning"] = data_dic["Column"].map(
    kepler_meanings
)
print(data_dic)
print()





print("-----------------------------------------")
undocu = data_dic[
    data_dic["Meaning"].isnull()
]
print("Columns still needing documentation:")
print(undocu["Column"].tolist())
print()






print("------------------------------------------")
ml_role = []
for column in df_clean.columns:
    if column in id_col:
        role = "Do not use directly"
    elif column == target_col:
        role = "Target"
    else:
        role = "Potential feature"
    ml_role.append(role)
data_dic["ML Role"] = ml_role
print(data_dic)
print()





print("======================================")
data_dic.to_csv(
    "data_dictionary.csv",
    index=False
)
print("Data dictionary saved successfully!")
print()





print("---------Missing values----------")
mis_values = df_clean.isnull().sum()
print("Missing values in each column:")
print(mis_values[mis_values > 0])
print()






print("-----------Missing percentage----------")
mis_percent = (
    df_clean.isnull().sum() / len(df_clean) * 100
).round(2)
print("Missing percentage:")
print(mis_percent[mis_percent > 0])
print()




print("-----------------------------------------------")
missing_report = pd.DataFrame()
missing_report["Column"] = df_clean.columns
missing_report["Missing Values"] = [
    df_clean[column].isnull().sum()
    for column in df_clean.columns
]
missing_report["Missing %"] = [
    round(df_clean[column].isnull().mean() * 100, 2)
    for column in df_clean.columns
]
missing_report = missing_report[
    missing_report["Missing Values"] > 0
]
missing_report = missing_report.sort_values(
    "Missing %",
    ascending=False
)
display(missing_report)
print()






print("------------------------------------------")
missing_report = missing_report.merge(
    data_dic[
        ["Column", "Meaning", "Category"]
    ],
    on="Column",
    how="left"
)
display(missing_report)
print()






print("-----------Missing level and decision---------------")
def mis_lvl(percent):
    if percent < 30:
        return "Low"
    elif percent < 50:
        return "Medium"
    elif percent < 80:
        return "High"
    else:
        return "Very High"
        
missing_report["Missing Level"] = [
    mis_lvl(percent)
    for percent in missing_report["Missing %"]
]
display(missing_report)
print()





print("---------------------------------------------------------")
dec = []
for index, row in missing_report.iterrows():
    column = row["Column"]
    perc = row["Missing %"]
    catg = row["Category"]

    if perc >= 80:
        decision = "Remove column"
    elif catg == "Numerical":
        decision = "Fill median"
    elif catg == "Categorical":
        decision = "Fill mode"
    else:
        decision = "Review"
        
    dec.append(decision)

missing_report["Decision"] = dec
display(
    missing_report[
        [
            "Column",
            "Missing %",
            "Meaning",
            "Category",
            "Decision"
        ]
    ]
)
print()






print("---------Saving Missing value decision in a csv file------------")
missing_report.to_csv(
    "missing_val_decision.csv",
    index=False
)
print("Missing-value decisions saved successfully!")
print()





print("------Data cleaning--------")
col_to_rem = missing_report[
    missing_report["Decision"] == "Remove column"
]["Column"].tolist()
print("Columns to remove:")
print(col_to_rem)
print()





print("------Dropping column-----------")
df_clean = df_clean.drop(
    columns=col_to_rem
)
print("Columns removed:", len(col_to_rem))
print()







print("------------------------------------------")
num_col = df_clean.select_dtypes(
    include=np.number
).columns
for column in num_col:

    if df_clean[column].isnull().sum() > 0:
        median_val = df_clean[column].median()

        df_clean[column] = df_clean[column].fillna(
            median_val
        )
print("Numerical missing values filled.")
print()





print("-----------------------------------------")
categ_col = df_clean.select_dtypes(
    include="object"
).columns

for column in categ_col:

    if df_clean[column].isnull().sum() > 0:
        
        mode_val = df_clean[column].mode()[0]
        df_clean[column] = df_clean[column].fillna(
            mode_val
        )
print("Categorical missing values filled.")
print()




print("------checking remaining missing values-------")
rema_mis = df_clean.isnull().sum()
rema_mis = rema_mis[
    rema_mis > 0
]
print("Remaining missing values:")

if len(rema_mis) == 0:
    print("No missing values")
else:
    print(rema_mis)

print()
print()




print("-------Final comparision----------")
print("Original dataset:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print()

print("Cleaned dataset:")
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])
print()

print(
    "Total missing values after cleaning:",
    df_clean.isnull().sum().sum()
)
print()




print("-----Saving the cleaned dataset---------")
df_clean.to_csv(
    "Data_cleaned.csv",
    index=False
)
print("Cleaned dataset saved successfully!")
print()

print("-----------cleaning done---------------")
print()
print()






print("------------ OUTLIER FLAG----------------------")
outlier_columns = [
    "koi_period",
    "koi_duration",
    "koi_depth",
    "koi_prad",
    "koi_teq",
    "koi_insol",
    "koi_model_snr",
    "koi_steff",
    "koi_slogg",
    "koi_srad",
    "koi_kepmag"
]

outlier_report = []

for column in outlier_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    flag_column = column + "_outlier"

    df_clean[flag_column] = (
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)
    ).astype(int)

    outlier_count = df_clean[flag_column].sum()

    outlier_report.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": outlier_count
    })

outlier_report = pd.DataFrame(outlier_report)

flag_columns = [
    column + "_outlier"
    for column in outlier_columns
]

df_clean["outlier_flag"] = (
    df_clean[flag_columns].max(axis=1)
)

print("Outlier flagging completed.")
print()

print("Total rows:", len(df_clean))
print(
    "Rows containing at least one outlier:",
    df_clean["outlier_flag"].sum()
)

print("\nOutlier Report:")
display(outlier_report)
print()










print("-------------------- HOST STAR analyze using kepid----------------------")

unique_host_stars = df_clean["kepid"].nunique()

print("Total observations:", len(df_clean))
print("Unique host stars:", unique_host_stars)
print()

host_counts = (
    df_clean
    .groupby("kepid")
    .size()
    .reset_index(name="KOI_Count")
)
print("Host-star observation counts:")
display(host_counts.head(10))

multi_observation_hosts = host_counts[
    host_counts["KOI_Count"] > 1
]

print(
    "Host stars with more than one observation:",
    len(multi_observation_hosts)
)

host_summary = (
    df_clean
    .groupby("kepid")
    .agg(
        KOI_Count=("kepoi_name", "count"),

        Confirmed=("koi_disposition",
                   lambda x: (x == "CONFIRMED").sum()),

        Candidate=("koi_disposition",
                   lambda x: (x == "CANDIDATE").sum()),

        False_Positive=("koi_disposition",
                        lambda x: (x == "FALSE POSITIVE").sum())
    )
    .reset_index()
)
print("\nDetailed host-star summary:")
display(host_summary.head(10))

largest_systems = host_summary.sort_values(
    "KOI_Count",
    ascending=False
)
print("\nHost stars with highest number of observations:")
display(largest_systems.head(10))
print()










print("------------------GROUP DATA BY HOST STAR------------------------")

from sklearn.model_selection import GroupShuffleSplit

target_column = "koi_disposition"

group_column = "kepid"

group_split = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_index, test_index = next(
    group_split.split(
        df_clean,
        df_clean[target_column],
        groups=df_clean[group_column]
    )
)
train_df = df_clean.iloc[train_index].copy()

test_df = df_clean.iloc[test_index].copy()

print("Host-star based splitting completed.")
print()

print("Original dataset:")
print("Rows:", len(df_clean))
print()

print("Training dataset:")
print("Rows:", len(train_df))
print("Unique host stars:", train_df["kepid"].nunique())
print()

print("Testing dataset:")
print("Rows:", len(test_df))
print("Unique host stars:", test_df["kepid"].nunique())
print()

train_hosts = set(train_df["kepid"])
test_hosts = set(test_df["kepid"])

common_hosts = train_hosts.intersection(test_hosts)

print("Host stars present in both datasets:",
      len(common_hosts))

if len(common_hosts) == 0:
    print("PASS: No host star is shared between train and test.")
else:
    print("WARNING: Host-star overlap detected.")








print("--------------- FINAL PREPARED DATASET_-----------------------")


print("FINAL DATASET ")

print("\n1. DATASET SHAPE")
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])


print("\n2. MISSING VALUES")

missing_total = df_clean.isnull().sum().sum()

print("Total missing values:", missing_total)

if missing_total == 0:
    print("PASS: No missing values found.")
else:
    print("WARNING: Missing values still exist.")


print("\n3. DUPLICATE ROWS")

duplicate_rows = df_clean.duplicated().sum()

print("Duplicate rows:", duplicate_rows)

if duplicate_rows == 0:
    print("PASS: No duplicate rows found.")
else:
    print("WARNING: Duplicate rows found.")


print("\n4. INFINITE VALUES")

numeric_columns = df_clean.select_dtypes(
    include=np.number
).columns

infinite_values = np.isinf(
    df_clean[numeric_columns]
).sum().sum()

print("Infinite values:", infinite_values)

if infinite_values == 0:
    print("PASS: No infinite values found.")
else:
    print("WARNING: Infinite values found.")



print("\n5. TARGET DISTRIBUTION")

print(
    df_clean["koi_disposition"].value_counts()
)



print("\n6. OUTLIER INFORMATION")
print(
    "Rows with at least one flagged outlier:",
    df_clean["outlier_flag"].sum()
)



print("\n7. HOST-STAR SPLIT VALIDATION")

train_hosts = set(train_df["kepid"])
test_hosts = set(test_df["kepid"])

common_hosts = train_hosts.intersection(test_hosts)
print(
    "Common host stars between train and test:",
    len(common_hosts)
)
if len(common_hosts) == 0:
    print("PASS: Train and test host stars do not overlap.")
else:
    print("WARNING: Host-star overlap detected.")

print()

print("-------------------------------------------------------------")

if (
    missing_total == 0
    and duplicate_rows == 0
    and infinite_values == 0
    and len(common_hosts) == 0
):
    print("FINAL VALIDATION: PASSED")
else:
    print("FINAL VALIDATION: CHECK WARNING")

print("---------------------------------------------------")
print()








#---------------------READ BELOW----------------------------
# Only for our team to download and save the dataset 
# please remove unnecessary wordings in the below code if you used in your code

print("---------- SHARING - PREPARED DATA---------------")


import os

#make sure to have a folder in these names to save .
os.makedirs("prepared_data", exist_ok=True)
os.makedirs("reports", exist_ok=True)


# 1. Save complete prepared dataset

df_clean.to_csv(
    "prepared_data.csv",
    index=False
)


# 2. Save training dataset

train_df.to_csv(
    "kepler_train_by_host.csv",
    index=False
)

# 3. Save testing dataset

test_df.to_csv(
    "kepler_test_by_host.csv",
    index=False
)

# 4. Save host-star summary

host_summary.to_csv(
    "rep_host_star_summary.csv",
    index=False
)


# 5. Save outlier report

outlier_report.to_csv(
    "rep_outlier_report.csv",
    index=False
)


# 6. Create final validation summary dataset

validation_summary = pd.DataFrame({
    "Check": [
        "Rows",
        "Columns",
        "Missing Values",
        "Duplicate Rows",
        "Infinite Values",
        "Unique Host Stars",
        "Flagged Outlier Rows",
        "Training Rows",
        "Testing Rows",
        "Common Train-Test Host Stars"
    ],

    "Result": [
        df_clean.shape[0],
        df_clean.shape[1],
        missing_total,
        duplicate_rows,
        infinite_values,
        df_clean["kepid"].nunique(),
        df_clean["outlier_flag"].sum(),
        len(train_df),
        len(test_df),
        len(common_hosts)
    ]
})

validation_summary.to_csv(
    "rep_final_validation.csv",
    index=False
)
print()
print()




print("-------------ALL FILES EXPORTED SUCCESSFULLY---------------------")


print("\nPrepared Data Files:")
print("1. prepared_data.csv")
print("2. kepler_train_by_host.csv")
print("3. kepler_test_by_host.csv")
print()

print("\nReport Files:")
print("4. rep_host_star_summary.csv")
print("5. rep_outlier_report.csv")
print("6. rep_final_validation.csv")
print()

print("\nMember 1 data preparation work is completed.")
```
